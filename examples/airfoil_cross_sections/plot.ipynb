```python
import pandas as pd
import plotly.graph_objects as go
```

```python
def scatter(x, y, *dfs, labels=None, symbols=None, sizes=None):
    """
    Parameters
    ----------
    x, y    : column names for axes
    *dfs    : one or more DataFrames
    labels  : list of legend labels (one per df)
    symbols : list of marker symbols (e.g. "circle", "square", "diamond")
    sizes   : list of marker sizes (px)
    """
    n = len(dfs)
    labels  = labels  or [None] * n
    symbols = symbols or [None] * n
    sizes   = sizes   or [10]   * n

    fig = go.Figure()
    for df, label, symbol, size in zip(dfs, labels, symbols, sizes):
        marker = dict(size=size)
        if symbol:
            marker["symbol"] = symbol
        fig.add_trace(
            go.Scatter(
                x=df[x],
                y=df[y],
                mode="markers",
                name=label,
                text=df["airfoil_name"],
                marker=marker,
                hovertemplate=f"<b>%{{text}}</b><br>{x}=%{{x}}<br>{y}=%{{y}}<extra></extra>",
            )
        )
    fig.update_layout(
        title=f"{y.upper()} vs. {x.upper()}",
        template="plotly_white",
        height=600,
        width=600,
        xaxis_title=x.upper(),
        yaxis_title=y.upper(),
        showlegend=any(l is not None for l in labels),
    )
    fig.show()
```

```python
df_skin_al_001 = pd.read_csv("results_skin_al_001.csv")
df_skin_al_001 = df_skin_al_001[df_skin_al_001["status"] == "ok"]
print(len(df_skin_al_001))
display(df_skin_al_001.head(5))
```
